## 2. Conexión y Extracción de Datos desde SQL Server

Configuramos la conexión a SQL Server y extraemos todas las tablas relevantes usando SQLAlchemy y pandas, asegurando manejo de errores y cierre de conexiones.

```python
# Configura los detalles de la conexión desde variables de entorno.
# Copia .env.example a .env y define los valores. Nunca escribas credenciales aquí.
import os

host     = os.getenv('DB_HOST')
port     = os.getenv('DB_PORT', '1433')
database = os.getenv('DB_NAME')

if not all([host, database]):
    raise RuntimeError(
        "Faltan variables de entorno: define DB_HOST y DB_NAME. Ver .env.example"
    )

try:
    # Crea la cadena de conexión
    url = (
        'mssql+pyodbc://@{host}:{port}/{db}'
        '?trusted_connection=yes&driver=SQL+Server'
    ).format(host=host, port=port, db=database)

    # Crear conexion con base de datos
    engine = create_engine(url)
    print("Conexion a la base de datos realizada")

except Exception as e:
    print('Error:', e)
```

```python

# Función auxiliar para extracción segura
def extraer_tabla(query, engine, nombre_tabla):
    try:
        with engine.connect() as connection:
            df = pd.read_sql_query(text(query), connection)
            print(f"Tabla '{nombre_tabla}' extraída: {df.shape}")
            return df
    except Exception as e:
        print(f"Error extrayendo {nombre_tabla}: {e}")
        return pd.DataFrame()
```

```python

# Consultas SQL
query_desercion = """
SELECT Identificacion, Periodo, Reintegro, Periodo_Siguiente, EX_Periodo_Siguiente_Normal, nuevo_periodo, Status, Tipo_Salto, Estado_Alumno, Tipo_estado_alumno, Modalidad, Programa, Semestre_SINU, ciclo, año, AÑO_MEN, Estado_Alumno, anio_grado, Genero, Estado_Pago, RANGO_EDAD, RANGO_SALARIO, ESTA_TRABAJANDO, METODO_FINANCIAMIENTO, ZONA_RESIDENCIA, REGIMEN_SISTEMA_SALUD, PERTENECE_GRUPO_ETNICO, GRUPO_ETNICO, TIENE_DISCAPACIDAD, DISCAPACIDAD, LGBTIQ, ciudad, localidad_normalizada
FROM academico.historial_academico
"""
```

```python

query_geo = """
SELECT Identificacion, barrio, departamento, pais
FROM geo.ubicacion_estudiante
"""
```

```python

query_becas = """
SELECT Periodo, NOMBRE_GRUPO, IDENTIFICACION as Identificacion, NOMBRE_CONCEPTO, NOMBRE_CAUSA_NOTA
FROM becas.descuentos_beca
"""
```

```python

query_materias = """
SELECT IDENTIFICACION as Identificacion, [MATERIAS INSCRITAS], [MATERIAS APROBADAS], COD_PERIODO as Periodo, Porcentaje_aprobacion
FROM academico.aprobacion_materias
"""
```

```python

query_finanza = """
SELECT Periodo, Identificacion, TOTAL
FROM financiera.cartera
"""
```

```python

query_permanencia = """
SELECT *
FROM plataforma.permanencia
"""
```

```python

# Extracción de tablas
historial_academico = extraer_tabla(query_desercion, engine, "Desercion")
localizacion = extraer_tabla(query_geo, engine, "Geolocalizacion")
becas = extraer_tabla(query_becas, engine, "Becas")
materias = extraer_tabla(query_materias, engine, "Materias")
finanza = extraer_tabla(query_finanza, engine, "Finanza")
permanencia = extraer_tabla(query_permanencia, engine, "plataforma.permanencia")
```

## 3. Unión y Limpieza de Tablas (Merges y Preprocesamiento)

Realizamos los merges entre las tablas extraídas, eliminamos duplicados, estandarizamos nombres de columnas y filtramos periodos según criterios de negocio.

```python
# Copia de seguridad
df = historial_academico.copy()
```

```python

# Merge con localización
df1 = df.merge(localizacion, on='Identificacion', how='left')
```

```python

# Becas: eliminar duplicados por Identificacion y Periodo
becas = becas[~becas.duplicated(subset=['Identificacion', 'Periodo'], keep='last')]
df2 = df1.merge(becas, on=['Identificacion', 'Periodo'], how='left')
```

```python

# Materias
df3 = df2.merge(materias, on=['Identificacion', 'Periodo'], how='left')
```

```python

# Finanza: eliminar duplicados por Identificacion y Periodo
finanza = finanza[~finanza.duplicated(subset=['Identificacion', 'Periodo'], keep='last')]
df4 = df3.merge(finanza, on=['Identificacion', 'Periodo'], how='left')
```

```python

# Filtrado de periodos en permanencia
periodos_a_excluir = [
    '24I01', '24I02', '24I03', '24I04', '24I05', '24I06',
    '24I10', '24I11', '24I12', '24I13',
    '25I01', '25I02', '25I03', '25I04', '25I11', '25I12', '25I13'
]
permanencia = permanencia[~permanencia['cod_periodo'].isin(periodos_a_excluir)]
```

```python

# Obtener último periodo por estudiante
ultimo_periodo = permanencia.groupby('num_identificacion')['cod_periodo'].max().reset_index()
ultimo_periodo.rename(columns={'cod_periodo': 'ultimo_cod_periodo'}, inplace=True)
permanenciaa = permanencia.merge(ultimo_periodo, on='num_identificacion', how='left')
permanenciaa = permanenciaa[permanenciaa['cod_periodo'] == permanenciaa['ultimo_cod_periodo']]
permanenciaa = permanenciaa.drop_duplicates(subset=['num_identificacion', 'curso', 'cod_periodo'])
permanenciaa = permanenciaa.drop(columns=['ultimo_cod_periodo'])
permanenciaa = permanenciaa.sort_values(by=['num_identificacion', 'curso']).reset_index(drop=True)
permanenciaa = permanenciaa.rename(columns={'num_identificacion': 'Identificacion', 'cod_periodo': 'Periodo'})
```

```python

# Eliminar columnas innecesarias
cols_drop = [
    'SEMESTRE', 'NOM_UNIDAD', 'NOM_DEPENDENCIA', 'CRE_PROGRAMA',
    'CREDITOS_INSCRITOS', 'Materias_BU', 'Materias_B1', 'Materias_B2',
    'nivelado_en_materias', 'itemtype', 'itemmodule', 'itemname',
    'nombrecategoria', 'peso', 'finalgrade', 'estadoactividad',
    'estudiantes', 'curso', '#ASISTENCIAS'
]
permanenciaa = permanenciaa.drop(columns=[c for c in cols_drop if c in permanenciaa.columns], errors='ignore')
permanenciaa = permanenciaa.drop_duplicates(subset=['Identificacion', 'Periodo'])
```

```python

# Merge final
df5 = df4.merge(permanenciaa, on=['Identificacion', 'Periodo'], how='left')
```

```python

# Estandarización de nombres de columnas
df5.columns = df5.columns.str.strip().str.upper().str.replace(' ', '_')
```

## 5. Ingeniería de Características

Creamos nuevas variables relevantes, como indicadores binarios y transformaciones de columnas, para mejorar la capacidad predictiva del modelo.

```python
print(df5.columns.tolist())
```

```python
# Estandarización de nombres de columnas
df5.columns = df5.columns.str.strip().str.upper().str.replace(' ', '_')
```

```python
# Estandarización de nombres de columnas
df4.columns = df4.columns.str.strip().str.upper().str.replace(' ', '_')
```

```python
print(df5.columns.tolist())
```

```python
# =============================================================================
# 4. INGENIERÍA DE CARACTERÍSTICAS AVANZADA (TENDENCIAS)
# =============================================================================
# Asegúrate de que las columnas están en el formato correcto
df4['PERIODO'] = df4['PERIODO'].astype(str)
df4['PORCENTAJE_APROBACION'] = pd.to_numeric(df4['PORCENTAJE_APROBACION'], errors='coerce')
df4['ESTADO_PAGO'] = df4['ESTADO_PAGO'].astype(str)

# Ordenamos los datos para asegurar que los cálculos históricos sean correctos
df_historial = df4.sort_values(by=['IDENTIFICACION', 'PERIODO'])
grouped = df_historial.groupby('IDENTIFICACION')

# --- Característica 1: Tendencia del Rendimiento Académico ---
df_historial['TENDENCIA_APROBACION'] = grouped['PORCENTAJE_APROBACION'].diff()

# --- Característica 2: Historial de Pagos en Mora ---
df_historial['EN_MORA'] = (df_historial['ESTADO_PAGO'].str.strip().str.upper() == 'NO PAGO').astype(int)
df_historial['VECES_EN_MORA'] = grouped['EN_MORA'].cumsum()

# --- Creación del DataFrame final (df5) con el último registro de cada estudiante ---
# (Aquí iría tu lógica para filtrar y obtener el último registro de df4, por ejemplo, usando 'permanencia')
# Asumimos que 'df5' es el resultado final con un registro por estudiante.
# Luego, unimos las nuevas características.
nuevas_features = df_historial[['IDENTIFICACION', 'PERIODO', 'TENDENCIA_APROBACION', 'VECES_EN_MORA']]
df5 = df5.merge(nuevas_features, on=['IDENTIFICACION', 'PERIODO'], how='left')

# --- Característica 3: Bandera de no aprobación ---
df5['FLAG_NO_APROBO_NADA'] = np.where(df5['PORCENTAJE_APROBACION'] == 0, 1, 0)
df5['PORCENTAJE_APROBACION'] = df5['PORCENTAJE_APROBACION'].replace(0, np.nan)
print("Nuevas características de tendencia y banderas creadas.")

# --- Creación de la variable objetivo ---
df5['TARGET_DESERCION'] = df5['STATUS'].apply(lambda x: 1 if str(x).strip().lower() == 'desercion' else 0)
```

## 6. Selección y Preprocesamiento de Variables

Definimos las variables numéricas y categóricas, y construimos pipelines de preprocesamiento con imputación y codificación adecuada usando ColumnTransformer.

```python
# =============================================================================
# 5. SELECCIÓN DE VARIABLES Y CONSTRUCCIÓN DEL PIPELINE
# =============================================================================
columnas_numericas = [
    'SEMESTRE_SINU', 'MATERIAS_INSCRITAS', 'MATERIAS_APROBADAS',
    'PORCENTAJE_APROBACION', 'TOTAL', 'FLAG_NO_APROBO_NADA',
    'TENDENCIA_APROBACION', 'VECES_EN_MORA'
]
columnas_categoricas = [
    'TIPO_SALTO', 'MODALIDAD', 'GENERO', 'ESTADO_PAGO', 'RANGO_EDAD',
    'RANGO_SALARIO', 'ESTA_TRABAJANDO', 'METODO_FINANCIAMIENTO',
    'ZONA_RESIDENCIA', 'REGIMEN_SISTEMA_SALUD', 'ASISTENCIA'
]
```

```python

# Nos aseguramos de que solo usamos las columnas que existen en el df5 final
features = [col for col in (columnas_numericas + columnas_categoricas) if col in df5.columns]
X = df5[features]
y = df5['TARGET_DESERCION']
```

```python

# Definimos las columnas que realmente se usarán en el preprocesador
final_num_cols = [col for col in columnas_numericas if col in X.columns]
final_cat_cols = [col for col in columnas_categoricas if col in X.columns]

numeric_transformer = Pipeline(steps=[('imputer', SimpleImputer(strategy='median')), ('scaler', StandardScaler())])
categorical_transformer = Pipeline(steps=[('imputer', SimpleImputer(strategy='most_frequent')), ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))])
preprocessor = ColumnTransformer(transformers=[
    ('num', numeric_transformer, final_num_cols),
    ('cat', categorical_transformer, final_cat_cols)
])
```

## 7. División de Datos en Entrenamiento y Prueba

Dividimos el dataset en conjuntos de entrenamiento y prueba de forma estratificada para asegurar la representatividad de la variable objetivo.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)
print(f"Datos divididos: {len(X_train)} para entrenamiento, {len(X_test)} para prueba.")
```

## 8. Optimización de Hiperparámetros con GridSearchCV

Utilizamos GridSearchCV sobre un pipeline con XGBoost para encontrar los mejores hiperparámetros, usando una muestra estratificada del set de entrenamiento.

```python
xgb_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', xgb.XGBClassifier(random_state=42, use_label_encoder=False, eval_metric='logloss'))
])
```

```python

# Usar una muestra estratificada para acelerar la búsqueda
splitter = StratifiedShuffleSplit(n_splits=1, train_size=0.15, random_state=42)
for train_index, _ in splitter.split(X_train, y_train):
    X_sample, y_sample = X_train.iloc[train_index], y_train.iloc[train_index]

param_grid_xgb = {
    'classifier__n_estimators': [100, 200],
    'classifier__max_depth': [5, 7],
    'classifier__learning_rate': [0.05, 0.1]
}
grid_search = GridSearchCV(
    estimator=xgb_pipeline,
    param_grid=param_grid_xgb,
    cv=3,
    scoring='f1',
    n_jobs=-1,
    verbose=2
)
print("Iniciando la búsqueda de los mejores hiperparámetros...")
grid_search.fit(X_sample, y_sample)
print("Mejores parámetros encontrados:", grid_search.best_params_)
```

## 9. Balanceo de Clases con SMOTE

Aplicamos SMOTE al set de entrenamiento para balancear la variable objetivo antes del entrenamiento final del modelo.

```python
print("Preprocesando y aplicando SMOTE al set de entrenamiento completo...")
X_train_processed = preprocessor.fit_transform(X_train)
smote = SMOTE(random_state=42)
X_train_resampled, y_train_resampled = smote.fit_resample(X_train_processed, y_train)
print("Balanceo con SMOTE completado.")
```

## 10. Entrenamiento del Modelo XGBoost Final

Entrenamos el modelo XGBoost con los mejores hiperparámetros y el set de entrenamiento balanceado, asegurando reproducibilidad.

```python
best_params = grid_search.best_params_
final_params = {key.replace('classifier__', ''): value for key, value in best_params.items()}
final_classifier = xgb.XGBClassifier(**final_params, random_state=42, use_label_encoder=False, eval_metric='logloss')

print("Re-entrenando el mejor clasificador con todos los datos balanceados...")
final_classifier.fit(X_train_resampled, y_train_resampled)
print("Re-entrenamiento completado.")
```

## 11. Búsqueda del Mejor Umbral de Decisión

Calculamos el umbral óptimo para la clasificación binaria maximizando el F1-score sobre el set de prueba.

```python
print("Buscando el mejor umbral de decisión...")
X_test_processed = preprocessor.transform(X_test)
y_proba = final_classifier.predict_proba(X_test_processed)[:, 1]
thresholds = np.arange(0.1, 0.6, 0.01)
f1_scores = [f1_score(y_test, (y_proba >= t).astype(int)) for t in thresholds]
best_threshold = thresholds[np.argmax(f1_scores)]
print(f"Mejor umbral encontrado para maximizar F1-Score: {best_threshold:.2f}")
```

## 12. Evaluación del Modelo: Métricas y Visualizaciones

Evaluamos el modelo con métricas como classification_report, matriz de confusión, curva ROC y AUC, e importancia de variables.

```python
# Predicción final con umbral óptimo
y_pred_final = (y_proba >= best_threshold).astype(int)
```

```python

# Reporte de clasificación y matriz de confusión
print(f"--- Reporte de Clasificación Final (Umbral Óptimo {best_threshold:.2f}) ---")
print(classification_report(y_test, y_pred_final))
cm = confusion_matrix(y_test, y_pred_final)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='viridis')
plt.title(f'Matriz de Confusión - Modelo Final (Umbral {best_threshold:.2f})')
plt.ylabel('Etiqueta Real')
plt.xlabel('Etiqueta Predicha')
plt.show()
```

```python

# Curva ROC y AUC
auc = roc_auc_score(y_test, y_proba)
fpr, tpr, _ = roc_curve(y_test, y_proba)
plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, label=f'XGBoost Optimizado (AUC = {auc:.4f})')
plt.plot([0, 1], [0, 1], 'r--')
plt.title('Curva ROC del Modelo Final')
plt.xlabel('Tasa de Falsos Positivos')
plt.ylabel('Tasa de Verdaderos Positivos')
plt.legend()
plt.grid()
plt.show()
```

```python

# Importancia de variables
ohe_feature_names = preprocessor.named_transformers_['cat'].named_steps['onehot'].get_feature_names_out(columnas_categoricas)
final_feature_names = columnas_numericas + list(ohe_feature_names)
importancia = pd.Series(final_classifier.feature_importances_, index=final_feature_names).sort_values(ascending=False)
plt.figure(figsize=(12, 8))
sns.barplot(x=importancia.head(15), y=importancia.head(15).index)
plt.title('Top 15 Variables más importantes - Modelo Final Optimizado')
plt.xlabel('Importancia')
plt.ylabel('Variable')
plt.show()
```

## 15. Función Profesional de Predicción para Nuevos Datos

Implementamos una función robusta para cargar el pipeline y el modelo, y realizar predicciones sobre nuevos datos con el umbral óptimo.

```python
def predecir_desercion(nuevos_datos, preprocessor_path='preprocessor.pkl', model_path='xgb_classifier.pkl', umbral=best_threshold):
    """
    Carga los componentes del modelo y predice la deserción para nuevos datos.
    :param nuevos_datos: DataFrame con los mismos 'features' originales.
    :param preprocessor_path: Ruta al preprocesador guardado.
    :param model_path: Ruta al modelo guardado.
    :param umbral: Umbral de decisión para clasificar como desertor.
    :return: Un array con las predicciones (1 para deserción, 0 para no).
    """
    preproc = joblib.load(preprocessor_path)
    model = joblib.load(model_path)
    nuevos_datos_proc = preproc.transform(nuevos_datos)
    probabilidades = model.predict_proba(nuevos_datos_proc)[:, 1]
    predicciones = (probabilidades >= umbral).astype(int)
    return predicciones

# Ejemplo de uso:
muestra_nuevos_datos = X_test.head(5)
predicciones_ejemplo = predecir_desercion(muestra_nuevos_datos)
print("Predicciones para 5 estudiantes de prueba:", predicciones_ejemplo)
```

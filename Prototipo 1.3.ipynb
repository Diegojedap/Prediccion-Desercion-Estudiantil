```python
# =============================================================================
# 5. DIVISIÓN DE DATOS Y BALANCEO DE CLASES (SMOTE)
# =============================================================================
# --- Separar características (X) y objetivo (y) ---
X = df5_binarias.drop('TARGET_DESERCION', axis=1)
y = df5_binarias['TARGET_DESERCION']
```

```python

# --- Dividir en conjuntos de entrenamiento y prueba (80/20) ---
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)

print(f"Tamaño del conjunto de entrenamiento: {X_train.shape}")
print(f"Tamaño del conjunto de prueba: {X_test.shape}")
print("\nDistribución de clases en entrenamiento (original):")
print(y_train.value_counts())
```

```python

# --- Aplicar SMOTE para balancear clases en el conjunto de entrenamiento ---
# Este proceso puede consumir bastante memoria si X_train es muy grande
print("\nAplicando SMOTE...")
smote = SMOTE(random_state=42)
X_train_resampled, y_train_resampled = smote.fit_resample(X_train, y_train)



print("\nDistribución de clases en entrenamiento (con SMOTE):")
print(y_train_resampled.value_counts())
print("Tamaño del set de entrenamiento remuestreado:", X_train_resampled.shape)
```

```python


# =============================================================================
# 6. OPTIMIZACIÓN DEL MODELO CON UNA MUESTRA DE DATOS
# =============================================================================
# --- Estrategia 1: Usar una muestra para acelerar la búsqueda de hiperparámetros ---

```

```python

# Se crea una muestra del 15% del total de datos remuestreados para la búsqueda.
# Esto es mucho más rápido que usar los 1.2 millones de registros.
print("\nCreando una muestra para la optimización...")
splitter = StratifiedShuffleSplit(n_splits=1, train_size=0.15, random_state=42)
for train_index, _ in splitter.split(X_train_resampled, y_train_resampled):
    X_sample = X_train_resampled.iloc[train_index]
    y_sample = y_train_resampled.iloc[train_index]

print(f"Tamaño de la muestra para GridSearchCV: {X_sample.shape}")
```

```python

# --- Búsqueda de hiperparámetros con GridSearchCV sobre la muestra ---
from sklearn.model_selection import GridSearchCV

param_grid = {
    'n_estimators': [100, 200],      # Número de árboles
    'max_depth': [10, 20, None],     # Profundidad máxima
    'min_samples_split': [2, 5],     # Mínimo de muestras para dividir un nodo
    'min_samples_leaf': [1, 2]       # Mínimo de muestras en una hoja
}

print("\nIniciando optimización de hiperparámetros sobre la muestra (esto puede tardar)...")
grid_search = GridSearchCV(estimator=RandomForestClassifier(random_state=42),
                           param_grid=param_grid,
                           cv=3,
                           n_jobs=-1,
                           verbose=2,
                           scoring='f1') # Optimizar para la métrica F1

grid_search.fit(X_sample, y_sample)

print("\nMejores parámetros encontrados:", grid_search.best_params_)

```

```python

# =============================================================================
# 7. ENTRENAMIENTO Y EVALUACIÓN DEL MODELO FINAL OPTIMIZADO
# =============================================================================
# --- Entrenar el modelo final UNA SOLA VEZ con los mejores parámetros y TODOS los datos ---
print("\nEntrenando el modelo final optimizado con TODOS los datos de entrenamiento...")
```

```python

# Se crea un nuevo clasificador con la mejor configuración encontrada
best_rf_final = RandomForestClassifier(**grid_search.best_params_, 
                                     random_state=42, 
                                     n_jobs=-1)
```

```python

# Se entrena con el conjunto de datos remuestreado completo (los 1.2 millones de filas)
best_rf_final.fit(X_train_resampled, y_train_resampled)
print("Entrenamiento final completado.")
```

```python

# --- Evaluación exhaustiva del modelo final en el conjunto de prueba ---
print("\n--- EVALUACIÓN FINAL DEL MODELO OPTIMIZADO ---")
y_pred_final = best_rf_final.predict(X_test)
```

```python

# Reporte de Clasificación
print("\nReporte de Clasificación (Modelo Optimizado):")
print(classification_report(y_test, y_pred_final))
```

```python

# Matriz de Confusión
print("\nMatriz de Confusión (Modelo Optimizado):")
cm_final = confusion_matrix(y_test, y_pred_final)
plt.figure(figsize=(8, 6))
sns.heatmap(cm_final, annot=True, fmt='d', cmap='Greens', cbar=False)
plt.title('Matriz de Confusión - Random Forest Optimizado')
plt.ylabel('Etiqueta Real')
plt.xlabel('Etiqueta Predicha')
plt.show()
```

```python
# Curva ROC y AUC
print("\nCalculando Curva ROC y AUC...")

y_proba_final = best_rf_final.predict_proba(X_test)[:, 1]
auc_final = roc_auc_score(y_test, y_proba_final)
print(f"Área Bajo la Curva (AUC) - Modelo Optimizado: {auc_final:.4f}")

fpr, tpr, _ = roc_curve(y_test, y_proba_final)
plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, label=f'Random Forest Optimizado (AUC = {auc_final:.4f})')
plt.plot([0, 1], [0, 1], 'r--', label='Clasificador Aleatorio')
plt.xlabel('Tasa de Falsos Positivos')
plt.ylabel('Tasa de Verdaderos Positivos')
plt.title('Curva ROC - Modelo Optimizado')
plt.legend()
plt.grid()
plt.show()
```
